```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Load the raw telemetry dataset
CSV_PATH = "/content/live_telemetry.csv"

df = pd.read_csv(CSV_PATH)

print("Dataset loaded successfully.")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
# ============================================
# CELL 2 — Schema & Basic Dataset Integrity
# ============================================

print("=" * 70)
print("DATASET SHAPE")
print("=" * 70)
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for i, col in enumerate(df.columns, start=1):
    print(f"{i:2d}. {col}")

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(df.dtypes)

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(df.head())

print("\n" + "=" * 70)
print("LAST 5 ROWS")
print("=" * 70)

display(df.tail())
```

```python
# ============================================
# CELL 3 — Complete Data Quality Audit
# ============================================

print("=" * 80)
print("MISSING VALUES")
print("=" * 80)

missing = df.isna().sum()
missing_pct = (missing / len(df) * 100).round(3)

missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_percentage": missing_pct
})

display(missing_report[missing_report["missing_count"] > 0]
        .sort_values("missing_count", ascending=False))


print("\n" + "=" * 80)
print("DUPLICATE ROWS")
print("=" * 80)

duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count:,}")


print("\n" + "=" * 80)
print("COLLECTION ERRORS")
print("=" * 80)

# Treat both NaN and empty-string representations as "no error"
if "collection_errors" in df.columns:
    collection_errors = df["collection_errors"].fillna("").astype(str).str.strip()

    actual_errors = collection_errors[
        (collection_errors != "") &
        (collection_errors.str.lower() != "nan")
    ]

    print(f"Rows with actual collection errors: {len(actual_errors):,}")

    if len(actual_errors) > 0:
        print("\nError values:")
        display(actual_errors.value_counts().head(20))


print("\n" + "=" * 80)
print("UNIQUE VALUES IN CATEGORICAL COLUMNS")
print("=" * 80)

categorical_columns = [
    "vm_id",
    "workload_profile",
    "phase",
    "workload_active"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))


print("\n" + "=" * 80)
print("NUMERIC SUMMARY")
print("=" * 80)

numeric_cols = df.select_dtypes(include=np.number).columns

display(
    df[numeric_cols]
    .describe()
    .T
    .round(3)
)
```

```python
# ============================================
# CELL 4 — Temporal Quality & Sampling Analysis
# ============================================

# Convert timestamp to timezone-aware datetime
df["timestamp"] = pd.to_datetime(df["timestamp"], utc=True)

print("=" * 80)
print("TIMESTAMP RANGE")
print("=" * 80)

print(f"Start: {df['timestamp'].min()}")
print(f"End  : {df['timestamp'].max()}")
print(f"Duration: {df['timestamp'].max() - df['timestamp'].min()}")


print("\n" + "=" * 80)
print("OVERALL TIME ORDER")
print("=" * 80)

print(f"Rows in chronological order: {df['timestamp'].is_monotonic_increasing}")

# Overall timestamp gaps
overall_gaps = df["timestamp"].diff().dt.total_seconds().dropna()

print("\nOverall timestamp gap statistics:")
display(
    overall_gaps.describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    ).to_frame("seconds").round(4)
)


print("\n" + "=" * 80)
print("PER-VM SAMPLING INTERVAL ANALYSIS")
print("=" * 80)

for vm in sorted(df["vm_id"].unique()):

    vm_df = df[df["vm_id"] == vm].sort_values("timestamp").copy()

    vm_gaps = (
        vm_df["timestamp"]
        .diff()
        .dt.total_seconds()
        .dropna()
    )

    print(f"\n{vm}")
    print(f"Rows: {len(vm_df):,}")
    print(f"Start: {vm_df['timestamp'].min()}")
    print(f"End:   {vm_df['timestamp'].max()}")

    if len(vm_gaps) > 0:
        print(
            f"Sampling interval → "
            f"median={vm_gaps.median():.4f}s, "
            f"mean={vm_gaps.mean():.4f}s, "
            f"min={vm_gaps.min():.4f}s, "
            f"max={vm_gaps.max():.4f}s"
        )

        print("Percentiles:")
        print(
            vm_gaps.quantile(
                [0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
            ).round(4)
        )


print("\n" + "=" * 80)
print("EXPERIMENT DURATION")
print("=" * 80)

experiment_summary = (
    df.groupby("experiment_id")
      .agg(
          vm_count=("vm_id", "nunique"),
          row_count=("vm_id", "size"),
          start_time=("timestamp", "min"),
          end_time=("timestamp", "max")
      )
)

experiment_summary["duration_seconds"] = (
    experiment_summary["end_time"] -
    experiment_summary["start_time"]
).dt.total_seconds()

display(
    experiment_summary[
        ["vm_count", "row_count", "duration_seconds"]
    ].describe().round(3)
)
```

```python
# ============================================
# CELL 5 — Workload-wise Memory Analysis
# ============================================

print("=" * 80)
print("MEMORY USAGE BY WORKLOAD PROFILE")
print("=" * 80)

workload_memory = (
    df.groupby("workload_profile")["guest_memory_used_mib"]
      .agg([
          "count",
          "mean",
          "median",
          "std",
          "min",
          "max"
      ])
      .round(3)
)

display(workload_memory)


print("\n" + "=" * 80)
print("MEMORY USAGE PERCENTILES BY WORKLOAD")
print("=" * 80)

percentiles = (
    df.groupby("workload_profile")["guest_memory_used_mib"]
      .quantile([0.05, 0.25, 0.50, 0.75, 0.95])
      .unstack()
)

percentiles.columns = [
    "P05",
    "P25",
    "P50",
    "P75",
    "P95"
]

display(percentiles.round(3))


# ------------------------------------------------
# Boxplot
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="workload_profile",
    y="guest_memory_used_mib"
)

plt.title("Guest Memory Usage by Workload Profile")
plt.xlabel("Workload Profile")
plt.ylabel("Guest Memory Used (MiB)")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------
# Workload-wise mean memory
# ------------------------------------------------

workload_mean = (
    df.groupby("workload_profile")["guest_memory_used_mib"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

workload_mean.plot(kind="bar")

plt.title("Average Guest Memory Usage by Workload")
plt.xlabel("Workload Profile")
plt.ylabel("Average Guest Memory Used (MiB)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------
# Peak memory by experiment
# ------------------------------------------------

experiment_peak = (
    df.groupby(["experiment_id", "workload_profile"])["guest_memory_used_mib"]
      .max()
      .reset_index()
)

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=experiment_peak,
    x="workload_profile",
    y="guest_memory_used_mib"
)

plt.title("Peak Guest Memory per Experiment by Workload")
plt.xlabel("Workload Profile")
plt.ylabel("Peak Guest Memory Used (MiB)")
plt.grid(axis="y", alpha=0.3)

plt.show()


print("\n" + "=" * 80)
print("PEAK MEMORY STATISTICS PER WORKLOAD")
print("=" * 80)

peak_summary = (
    experiment_peak.groupby("workload_profile")["guest_memory_used_mib"]
    .agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
    .round(3)
)

display(peak_summary)
```

```python
# ============================================
# CELL 6 — Memory Temporal Dynamics
# ============================================

print("=" * 80)
print("MEMORY CHANGE / TEMPORAL DYNAMICS")
print("=" * 80)

# Work on a time-sorted copy
temporal_df = df.sort_values(
    ["vm_id", "experiment_id", "timestamp"]
).copy()

# Memory change within each VM + experiment
temporal_df["memory_delta_mib"] = (
    temporal_df
    .groupby(["vm_id", "experiment_id"])["guest_memory_used_mib"]
    .diff()
)

# Memory change per second
temporal_df["memory_change_rate_mib_per_sec"] = (
    temporal_df["memory_delta_mib"] /
    temporal_df["elapsed_seconds_since_previous_sample"]
)

# Absolute memory change
temporal_df["abs_memory_delta_mib"] = (
    temporal_df["memory_delta_mib"].abs()
)

print("\nMemory delta statistics:")
display(
    temporal_df["memory_delta_mib"]
    .describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    )
    .to_frame("MiB")
    .round(4)
)

print("\nMemory change-rate statistics:")
display(
    temporal_df["memory_change_rate_mib_per_sec"]
    .replace([np.inf, -np.inf], np.nan)
    .describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    )
    .to_frame("MiB/sec")
    .round(4)
)


# ------------------------------------------------
# Workload-wise memory delta
# ------------------------------------------------

print("\n" + "=" * 80)
print("MEMORY DELTA BY WORKLOAD")
print("=" * 80)

delta_by_workload = (
    temporal_df
    .groupby("workload_profile")["memory_delta_mib"]
    .agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
    .round(4)
)

display(delta_by_workload)


# ------------------------------------------------
# Phase-wise memory delta
# ------------------------------------------------

print("\n" + "=" * 80)
print("MEMORY DELTA BY PHASE")
print("=" * 80)

delta_by_phase = (
    temporal_df
    .groupby("phase")["memory_delta_mib"]
    .agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
    .round(4)
)

display(delta_by_phase)


# ------------------------------------------------
# Large memory changes
# ------------------------------------------------

print("\n" + "=" * 80)
print("LARGE MEMORY CHANGES")
print("=" * 80)

large_changes = (
    temporal_df["abs_memory_delta_mib"]
    .dropna()
    .quantile([0.90, 0.95, 0.99, 0.995, 0.999])
)

display(large_changes.to_frame("absolute_memory_delta_mib").round(4))


# ------------------------------------------------
# Plot memory delta distribution
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.histplot(
    temporal_df["memory_delta_mib"].dropna(),
    bins=100,
    kde=True
)

plt.axvline(0, linestyle="--")

plt.title("Distribution of Consecutive Memory Changes")
plt.xlabel("Memory Change Between Samples (MiB)")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------
# Workload-wise delta boxplot
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=temporal_df,
    x="workload_profile",
    y="memory_delta_mib"
)

plt.axhline(0, linestyle="--")

plt.title("Memory Change Between Samples by Workload")
plt.xlabel("Workload Profile")
plt.ylabel("Memory Delta (MiB)")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------
# Example time-series: one experiment per
# memory-intensive workload
# ------------------------------------------------

selected_profiles = [
    "memory_burst",
    "memory_ramp",
    "mixed"
]

for profile in selected_profiles:

    profile_experiments = (
        temporal_df[
            temporal_df["workload_profile"] == profile
        ]["experiment_id"]
        .drop_duplicates()
    )

    if len(profile_experiments) == 0:
        continue

    experiment_id = profile_experiments.iloc[0]

    example = temporal_df[
        temporal_df["experiment_id"] == experiment_id
    ].sort_values("timestamp")

    plt.figure(figsize=(12, 5))

    for vm in example["vm_id"].unique():

        vm_example = example[
            example["vm_id"] == vm
        ]

        plt.plot(
            vm_example["timestamp"],
            vm_example["guest_memory_used_mib"],
            marker="o",
            markersize=3,
            label=vm
        )

    plt.title(
        f"Guest Memory Over Time — {profile}"
    )

    plt.xlabel("Time")
    plt.ylabel("Guest Memory Used (MiB)")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.xticks(rotation=30)

    plt.show()
```

```python
# ============================================
# CELL 7 — Sampling Anomalies & Continuity
# ============================================

print("=" * 80)
print("SHORT SAMPLING INTERVAL ANALYSIS")
print("=" * 80)

# Work from experiment/VM sorted data
continuity_df = df.sort_values(
    ["vm_id", "experiment_id", "timestamp"]
).copy()

continuity_df["time_gap_sec"] = (
    continuity_df
    .groupby(["vm_id", "experiment_id"])["timestamp"]
    .diff()
    .dt.total_seconds()
)

# ------------------------------------------------
# Gap distribution
# ------------------------------------------------

valid_gaps = continuity_df["time_gap_sec"].dropna()

print(f"Total valid within-experiment gaps: {len(valid_gaps):,}")

print("\nGap statistics:")
display(
    valid_gaps.describe(
        percentiles=[0.001, 0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99, 0.999]
    )
    .to_frame("seconds")
    .round(4)
)


# ------------------------------------------------
# Count unusual short gaps
# ------------------------------------------------

print("\n" + "=" * 80)
print("SHORT GAP COUNTS")
print("=" * 80)

for threshold in [1.0, 2.0, 4.0, 4.5, 5.0]:
    count = (valid_gaps < threshold).sum()
    percentage = count / len(valid_gaps) * 100

    print(
        f"Gap < {threshold:>4.1f} sec: "
        f"{count:>6,} ({percentage:.3f}%)"
    )


# ------------------------------------------------
# Count unusually long gaps
# ------------------------------------------------

print("\n" + "=" * 80)
print("LONG GAP COUNTS")
print("=" * 80)

for threshold in [6.0, 10.0, 30.0, 60.0]:
    count = (valid_gaps > threshold).sum()
    percentage = count / len(valid_gaps) * 100

    print(
        f"Gap > {threshold:>4.0f} sec: "
        f"{count:>6,} ({percentage:.3f}%)"
    )


# ------------------------------------------------
# Inspect short-gap rows
# ------------------------------------------------

short_gaps = continuity_df[
    continuity_df["time_gap_sec"] < 1.0
].copy()

print("\n" + "=" * 80)
print("SHORT GAP BREAKDOWN")
print("=" * 80)

print(f"Rows with gap < 1 second: {len(short_gaps):,}")

if len(short_gaps) > 0:

    print("\nBy VM:")
    display(
        short_gaps["vm_id"]
        .value_counts()
        .to_frame("count")
    )

    print("\nBy workload:")
    display(
        short_gaps["workload_profile"]
        .value_counts()
        .to_frame("count")
    )

    print("\nBy phase:")
    display(
        short_gaps["phase"]
        .value_counts()
        .to_frame("count")
    )

    print("\nExamples:")
    display(
        short_gaps[
            [
                "timestamp",
                "experiment_id",
                "vm_id",
                "workload_profile",
                "phase",
                "time_gap_sec",
                "guest_memory_used_mib"
            ]
        ]
        .head(20)
        .round(4)
    )


# ------------------------------------------------
# Check normal sampling window
# ------------------------------------------------

print("\n" + "=" * 80)
print("NORMAL ~5 SECOND SAMPLING WINDOW")
print("=" * 80)

normal_gaps = valid_gaps[
    (valid_gaps >= 4.5) &
    (valid_gaps <= 5.8)
]

print(
    f"Gaps between 4.5 and 5.8 sec: "
    f"{len(normal_gaps):,} "
    f"({len(normal_gaps)/len(valid_gaps)*100:.3f}%)"
)

print("\nNormal-gap statistics:")
display(
    normal_gaps.describe().to_frame("seconds").round(4)
)


# ------------------------------------------------
# Very long gaps
# ------------------------------------------------

long_gaps = continuity_df[
    continuity_df["time_gap_sec"] > 30
].copy()

print("\n" + "=" * 80)
print("LONG GAP EXAMPLES")
print("=" * 80)

print(f"Number of gaps > 30 sec: {len(long_gaps):,}")

if len(long_gaps) > 0:
    display(
        long_gaps[
            [
                "timestamp",
                "experiment_id",
                "vm_id",
                "workload_profile",
                "phase",
                "time_gap_sec"
            ]
        ]
        .head(20)
        .round(3)
    )
```

```python
# ============================================
# CELL 9 — Memory Autocorrelation & Lag Analysis
# ============================================

print("=" * 80)
print("MEMORY AUTOCORRELATION / LAG ANALYSIS")
print("=" * 80)

# IMPORTANT:
# Analyze each VM + experiment independently.
# We do NOT allow lags to cross experiment boundaries.

lag_df = (
    df.sort_values(
        ["vm_id", "experiment_id", "timestamp"]
    )
    .copy()
)

# Create lagged memory values within each experiment
for lag in range(1, 21):
    lag_df[f"memory_lag_{lag}"] = (
        lag_df
        .groupby(["vm_id", "experiment_id"])["guest_memory_used_mib"]
        .shift(lag)
    )


# ------------------------------------------------
# Correlation between current memory and lags
# ------------------------------------------------

lag_correlations = []

for lag in range(1, 21):

    valid = lag_df[
        [
            "guest_memory_used_mib",
            f"memory_lag_{lag}"
        ]
    ].dropna()

    correlation = valid[
        "guest_memory_used_mib"
    ].corr(
        valid[f"memory_lag_{lag}"]
    )

    lag_correlations.append({
        "lag": lag,
        "approx_time_seconds": lag * 5.15,
        "approx_time_minutes": lag * 5.15 / 60,
        "correlation": correlation,
        "samples": len(valid)
    })


lag_corr_df = pd.DataFrame(lag_correlations)

print("\nLag correlation table:")
display(
    lag_corr_df.round(4)
)


# ------------------------------------------------
# Plot lag correlation
# ------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    lag_corr_df["lag"],
    lag_corr_df["correlation"],
    marker="o"
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xticks(
    range(1, 21)
)

plt.title(
    "Guest Memory Autocorrelation Across Temporal Lags"
)

plt.xlabel(
    "Lag (samples)"
)

plt.ylabel(
    "Correlation with Current Memory"
)

plt.grid(alpha=0.3)

plt.show()


# ------------------------------------------------
# Spearman correlation as a robustness check
# ------------------------------------------------

print("\n" + "=" * 80)
print("SPEARMAN LAG CORRELATION")
print("=" * 80)

spearman_results = []

for lag in range(1, 21):

    valid = lag_df[
        [
            "guest_memory_used_mib",
            f"memory_lag_{lag}"
        ]
    ].dropna()

    correlation = valid[
        "guest_memory_used_mib"
    ].corr(
        valid[f"memory_lag_{lag}"],
        method="spearman"
    )

    spearman_results.append({
        "lag": lag,
        "approx_seconds": lag * 5.15,
        "spearman_correlation": correlation
    })

spearman_df = pd.DataFrame(
    spearman_results
)

display(
    spearman_df.round(4)
)


# ------------------------------------------------
# Workload-specific lag-1 correlation
# ------------------------------------------------

print("\n" + "=" * 80)
print("LAG-1 CORRELATION BY WORKLOAD")
print("=" * 80)

lag1_by_workload = []

for workload in sorted(
    lag_df["workload_profile"].unique()
):

    subset = lag_df[
        lag_df["workload_profile"] == workload
    ][
        [
            "guest_memory_used_mib",
            "memory_lag_1"
        ]
    ].dropna()

    corr = subset[
        "guest_memory_used_mib"
    ].corr(
        subset["memory_lag_1"]
    )

    lag1_by_workload.append({
        "workload_profile": workload,
        "lag1_correlation": corr,
        "samples": len(subset)
    })

lag1_workload_df = pd.DataFrame(
    lag1_by_workload
)

display(
    lag1_workload_df.round(4)
)
```

```python
# ============================================
# CELL 10 — Future Horizon Predictability
# ============================================

print("=" * 80)
print("FUTURE MEMORY PREDICTABILITY BY HORIZON")
print("=" * 80)

horizon_df = (
    df.sort_values(
        ["vm_id", "experiment_id", "timestamp"]
    )
    .copy()
)

for h in [1, 2, 3, 5, 10]:

    horizon_df[f"memory_future_{h}"] = (
        horizon_df
        .groupby(["vm_id", "experiment_id"])
        ["guest_memory_used_mib"]
        .shift(-h)
    )

results = []

for h in [1, 2, 3, 5, 10]:

    current = horizon_df["guest_memory_used_mib"]
    future = horizon_df[f"memory_future_{h}"]

    valid = pd.DataFrame({
        "current": current,
        "future": future
    }).dropna()

    errors = valid["future"] - valid["current"]

    mae = errors.abs().mean()
    rmse = np.sqrt((errors ** 2).mean())

    correlation = valid["current"].corr(
        valid["future"],
        method="spearman"
    )

    results.append({
        "horizon_steps": h,
        "approx_seconds": h * 5.15,
        "approx_minutes": h * 5.15 / 60,
        "spearman_current_vs_future": correlation,
        "persistence_MAE_MiB": mae,
        "persistence_RMSE_MiB": rmse,
        "mean_absolute_memory_change_MiB": errors.abs().mean(),
        "samples": len(valid)
    })

horizon_results = pd.DataFrame(results)

display(
    horizon_results.round(4)
)


# ------------------------------------------------
# Future memory correlation
# ------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    horizon_results["approx_seconds"],
    horizon_results["spearman_current_vs_future"],
    marker="o"
)

plt.xlabel("Prediction Horizon (seconds)")
plt.ylabel(
    "Spearman Correlation: Current vs Future Memory"
)

plt.title(
    "Memory Predictability Across Future Horizons"
)

plt.grid(alpha=0.3)

plt.show()


# ------------------------------------------------
# Persistence baseline MAE
# ------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    horizon_results["approx_seconds"],
    horizon_results["persistence_MAE_MiB"],
    marker="o"
)

plt.xlabel("Prediction Horizon (seconds)")
plt.ylabel("Persistence MAE (MiB)")

plt.title(
    "Naive Persistence Error Across Prediction Horizons"
)

plt.grid(alpha=0.3)

plt.show()


# ------------------------------------------------
# Workload-specific horizon analysis
# ------------------------------------------------

print("\n" + "=" * 80)
print("HORIZON ERROR BY WORKLOAD")
print("=" * 80)

workload_horizon_results = []

for workload in sorted(
    horizon_df["workload_profile"].unique()
):

    workload_data = horizon_df[
        horizon_df["workload_profile"] == workload
    ]

    for h in [1, 2, 3, 5, 10]:

        current = workload_data[
            "guest_memory_used_mib"
        ]

        future = workload_data[
            f"memory_future_{h}"
        ]

        valid = pd.DataFrame({
            "current": current,
            "future": future
        }).dropna()

        errors = valid["future"] - valid["current"]

        workload_horizon_results.append({
            "workload_profile": workload,
            "horizon_steps": h,
            "approx_seconds": h * 5.15,
            "MAE_MiB": errors.abs().mean(),
            "RMSE_MiB": np.sqrt(
                (errors ** 2).mean()
            ),
            "samples": len(valid)
        })

workload_horizon_df = pd.DataFrame(
    workload_horizon_results
)

display(
    workload_horizon_df.round(4)
)
```
